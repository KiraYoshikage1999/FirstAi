# Report for sale department


```python
sales = [120, 135, 100, 150, 90, 200, 179]

total = sum(sales)
average = total / len(sales)
maximum = max(sales)
minimum = min(sales)

report_data = {
    "Name": ["Total sum", "Average", "Max", "Min"],
    "Data": [total, average, maximum, minimum]
}

df_report = pd.DataFrame(report_data)

df_report
```

# Marks of Students


```python
students = [("Alex", 85), ("Maria", 92), ("John", 74), ("Olga", 60), ("Max", 48)]

scores = [student[1] for student in students]

average_score = sum(scores) / len(scores)
max_score = max(scores)
min_score = min(scores)
low_score_students = [student[0] for student in students if student[1] < 60]
df_students = pd.DataFrame({
    "Student": [name for name, score in students],
    "Score": scores,
    "Visualization": ["*" * (score // 10) for name, score in students]
})

display(df_students)
df_summary = pd.DataFrame({
    "Display": [
        "Average score",
        "Maximum score",
        "Minimum score",
        "Students with score < 60"
    ],
    "Value": [
        f"{average_score:.2f}",
        str(max_score),
        str(min_score),
        ", ".join(low_score_students) if low_score_students else "None"
    ]
})

print("--- Summary Statistics ---")
display(df_summary)
```
